```python
import pandas as pd

df = pd.read_csv('Titanic-Dataset(1).csv')
df.head()
```

```python
filtered = df[(df['Survived'] == 1) & (df['Sex'] == 'female')]
print(filtered.head())
```

```python
grouped = df.groupby('Pclass')
print(grouped.mean(numeric_only=True))
```

```python
print("Mean:\n", df.mean(numeric_only=True))
print("\nSum:\n", df.sum(numeric_only=True))
print("\nCount:\n", df.count())
print("\nMinimum:\n", df.min(numeric_only=True))
print("\nMaximum:\n", df.max(numeric_only=True))
```

```python
agg = df.groupby('Sex').agg({
    'Age': ['mean', 'min', 'max'],
    'Fare': ['sum', 'mean'],
    'Survived': ['mean', 'count']
})

print(agg)
```
