```python
# Factorial Program
num = int(input("Enter a number: "))
fact = 1

for i in range(1, num + 1):
    fact *= i

print("Factorial of", num, "is:", fact)
```

```python
# Table Program
num = int(input("Enter a number: "))

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

```python
# Odd or Even Program
num = int(input("Enter a number: "))

if num % 2 == 0:
    print(num, "is Even")
else:
    print(num, "is Odd")
```

```python
# Palindrome Program
num = int(input("Enter a number: "))
original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Simple Expert System (Weather Suggestion)
print("Answer with yes/no")

rain = input("Is it raining? ").lower()
cold = input("Is it cold? ").lower()

if rain == "yes" and cold == "yes":
    print("Advice: Wear a raincoat and warm clothes")
elif rain == "yes":
    print("Advice: Carry an umbrella")
elif cold == "yes":
    print("Advice: Wear warm clothes")
else:
    print("Advice: Enjoy the weather!")
```
