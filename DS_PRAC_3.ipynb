```python
import numpy as np

# Create 3x3 matrix
A = np.array([[1, 2, 3],
              [0, 1, 4],
              [5, 6, 0]])

# Compute determinant
det = np.linalg.det(A)

print("Matrix:\n", A)
print("Determinant:", det)
```

```python
import numpy as np

# Create same matrix
A = np.array([[1, 2, 3],
              [0, 1, 4],
              [5, 6, 0]])

# Find inverse
A_inv = np.linalg.inv(A)

print("Matrix:\n", A)
print("Inverse:\n", A_inv)
```

```python
import numpy as np

# Coefficient matrix
A = np.array([[2, 3],
              [4, 5]])

# Constant matrix
B = np.array([10, 20])

# Solve equations
solution = np.linalg.solve(A, B)

print("Solution (x, y):", solution)
```
